Task 7. Geographic analysis of Python developer compensation

```python
result_country = result.copy()
result_country.dropna(subset=['Country'], inplace=True)

result_country['Country']
```

```python
result_country.dropna(subset=['LanguageHaveWorkedWith'], inplace=True)
result_country['LanguageHaveWorkedWith']
```

```python
python_responders_with_country = result_country['LanguageHaveWorkedWith'].str.contains('Python', case=False, na=False)
result_country['LanguageHaveWorkedWithPython'] = python_responders_with_country
result_country['LanguageHaveWorkedWithPython']
```

```python
result_country = result_country[result_country['LanguageHaveWorkedWithPython'] == True]
result_country
```

```python
#fill empty values with median

result_country['ConvertedCompYearly'].info()
```

```python
fill_empty_values_median = result_country.copy()

result_country_median = fill_empty_values_median['ConvertedCompYearly'].median()
fill_empty_values_median['ConvertedCompYearly'] = fill_empty_values_median['ConvertedCompYearly'].fillna(result_country_median)
fill_empty_values_median['ConvertedCompYearly'].info()
```

```python
fill_mean = fill_empty_values_median.groupby('Country', as_index=False).agg(
    MeanCompensationByCountry=('ConvertedCompYearly', 'mean'),
    MedianCompensationByCountry=('ConvertedCompYearly', 'median')
).round(2)
fill_mean
```

```python
#without empty values - delete

delete_empty_values_compansation = result_country.copy()

delete_empty_values_compansation.dropna(subset=['ConvertedCompYearly'], inplace=True)
delete_empty_values_compansation['ConvertedCompYearly'].info()
```

```python
delete_empty = delete_empty_values_compansation.groupby('Country', as_index=False).agg(
    MeanCompensationByCountry=('ConvertedCompYearly', 'mean'),
    MedianCompensationByCountry=('ConvertedCompYearly', 'median')
).round(2)
delete_empty
```

Task 9. Analysis of Python popularity by age category

```python
result.dropna(subset=['LanguageHaveWorkedWith'], inplace=True)

new_python_responders= result['LanguageHaveWorkedWith'].str.contains('Python', case=False, na=False)
result['LanguageHaveWorkedWithPython'] = new_python_responders
result = result[result['LanguageHaveWorkedWithPython'] == True]

result['Age'].info()

unique_age_category = result['Age'].nunique()
unique_age_category

popul_python_age = result.groupby('Age', as_index=False).agg(
    TopPythonAge=('LanguageHaveWorkedWithPython', 'sum')
)

popul_python_age['PercOfAge'] = (((popul_python_age['TopPythonAge'])/(new_python_responders.sum()))*100).round(2)
popul_python_age.sort_values('PercOfAge', ascending=False)
```

Task 10. Analysis of industries among high-paid remote workers

```python
remote_result = result.copy()
remote_responders = remote_result[remote_result['RemoteWork'] == 'Remote']['RemoteWork']
remote_result['Remote'] = remote_responders
remote_result.dropna(subset=['Remote'], inplace=True)

percentile_75 = remote_result['ConvertedCompYearly'].quantile(0.75)
remote_responders_percentile_75 = remote_result[remote_result['ConvertedCompYearly'] > percentile_75]
popular_industries = remote_responders_percentile_75['Industry'].value_counts()

popular_industries
```
